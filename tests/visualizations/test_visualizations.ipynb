### Local classification

```python
classes_names = ["Negative", "Positive"]
sample = ["Great", "food", "but", "slow"]
concepts_labels = ["Sentiment", "Food", "Service"]
concepts_labels_classwise = {
    0: {0: "Neg sentiment", 1: "Service issues"},
    1: {0: "Pos sentiment", 1: "Food quality"},
}
concepts_activations_tokens = [
    [0.8, 0.2, 0.1],
    [0.1, 0.9, 0.0],
    [0.0, 0.1, 0.2],
    [0.2, 0.1, 0.9],
]
concepts_activations_classwise = {
    0: [[0.4, 0.5]],
    1: [[0.3, 0.7]],
}
concepts_activations_pooled = [
    [0.4, 0.5, 0.3],
]
concepts_importances = [
    [0.2, 0.1, 0.6],
    [0.7, 0.3, 0.1],
]
concepts_importances_classwise = {
    0: [0.2, 0.8],
    1: [0.8, 0.2],
}

# Element-wise activations + shared labels
print("Element-wise activations + shared labels")
plot_concepts(
    sample=sample,
    classes_names=classes_names,
    concepts_activations=concepts_activations_tokens,
    concepts_importances=concepts_importances,
    concepts_labels=concepts_labels,
)

# Pooled activations + shared labels
print("Pooled activations + shared labels")
plot_concepts(
    sample=sample,
    classes_names=classes_names,
    concepts_activations=concepts_activations_pooled,
    concepts_importances=concepts_importances,
    concepts_labels=concepts_labels,
)

# Pooled activations + class-wise labels
print("Pooled activations + class-wise labels")
plot_concepts(
    sample=sample,
    classes_names=classes_names,
    concepts_activations=concepts_activations_classwise,
    concepts_importances=concepts_importances_classwise,
    concepts_labels=concepts_labels_classwise,
)
```
