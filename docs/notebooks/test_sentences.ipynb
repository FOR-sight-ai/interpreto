```python
from transformers import AutoTokenizer

from interpreto import Granularity

# ----------------------------
# Example input text
# ----------------------------
raw_text = ["I like url.com this !! How are you? I am \n fine... This is ahah ?! Mr. Smith is here."]

# ----------------------------
# Load tokenizer
# ----------------------------
tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokenizer.add_special_tokens({"pad_token": "[PAD]"})  # Add a padding token if not present

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# ----------------------------
# Tokenize
# ----------------------------
inputs = tokenizer(
    raw_text,
    return_tensors="pt",
    padding=True,
    truncation=True,
)

print("Input IDs shape:", inputs["input_ids"].shape)
print()

# ----------------------------
# Show decoded tokens
# ----------------------------
sentence_i = 0
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][sentence_i])
print("Tokens:")
print(tokens)
print()

# ----------------------------
# SENTENCE granularity indices
# ----------------------------
gran = Granularity.SENTENCE
indices = gran.get_indices(inputs, tokenizer)


print("Sentence indices:")
print(indices)
print()

# ----------------------------
# Sentence decomposition (text)
# ----------------------------
decomposition = gran.get_decomposition(
    inputs,
    tokenizer=tokenizer,
    return_text=True,
    raw_text=raw_text,
)

print("Sentence decomposition:")
for i, sent in enumerate(decomposition[sentence_i]):
    print(f"Sentence {i}: {repr(sent)}")
```
